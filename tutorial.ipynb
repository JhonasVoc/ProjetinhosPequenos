```python
prints(1,2,3,4,5)
```

```python
def soma(*args):
    soma = 0
    for i in args:
        soma += i
    
    return soma
```

```python
soma(2,2,3,2,2,2,2)
```

```python
sum1 = soma(2,2,3,2,2,2,2)
print(sum1)
```
